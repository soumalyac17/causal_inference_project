```r
# Set seed for reproducibility
set.seed(123)

# Define the covariance matrix for X1, X2, X3
cov_matrix <- matrix(c(
  2,   1,  -1,
  1,   1, -0.5,
 -1, -0.5,  1
), nrow = 3, byrow = TRUE)

# Generate index number -- adding column index to X1 - X6
index <- seq(from = 1, to = 120, by = 1)

# Generate multivariate normal data for X1, X2, X3
X123 <- mvrnorm(n = 120, mu = c(0, 0, 0), Sigma = cov_matrix)

# Generate independent variables X4, X5, X6
X4 <- runif(120, min = -3, max = 3)
X5 <- rchisq(120, df = 1)
X6 <- rbinom(120, size = 1, prob = 0.5)

# Combine all variables into a data frame
df <- data.frame(
  index = index,
  X1 = X123[,1],
  X2 = X123[,2],
  X3 = X123[,3],
  X4 = X4,
  X5 = X5,
  X6 = X6
)

X = cbind(X123,X4,X5,X6)
colnames(X) = c('x1','x2','x3','x4','x5','x6')
k = ncol(X)

# Write the data frame to a CSV file
write.csv(df, "paper1_data.csv", row.names = FALSE)

#----- new data set with X1, X2, .. X6  ----------------
print("Preparing a new data set of 120 records with 6 varaibles X1, ..X6 and one index column, drawing from covariance matrix and normal distribution")
data <- read.csv("paper1_data.csv")
head(data)
tail(data)
```

```r
 df_sample = data.frame(data)  #Jiko Update
 round(colMeans(df_sample), 2)
 set.seed(7)
 # Specify the full sample size.
 N = nrow(df_sample)
 #print("Total Rows in df_sample : " + N)
 cat("Total Rows in df_sample : ", N)

 n1 = N/2
 n2 = N - n1

 # Generate a Selection Order Matrix.
 som_obs = som(data_frame = NULL, n_treat = 2, treat_sizes = c(n1, n2), method = 'SCOMARS', marginal_treat = rep((n2/N), N))
 print("Display first 10 rows of the Selection Order Matrix, generated")
 som_obs[1:10,]

 # Generate a treatment assignment given som_obs.
 f = fsm(data_frame = df_sample, SOM = som_obs, s_function = 'Dopt', eps = 0.0001, units_print = FALSE)

 # Generate augmented SOM.
 som_obs_augmented = f$som_appended

 # Display first 10 rows and first 7 columns of the augmented SOM.
 print("# Display first 10 rows and first 7 columns of the augmented SOM.")
 round(som_obs_augmented[1:10,1:8])  #Jiko Update

 # Augment df_sample with the treatment label.
 df_sample_aug = f$data_frame_allocated
 # Create a vector of observed treatment labels.
 Z_fsm_obs = df_sample_aug$Treat
 print("Printing z_fsm_obs generated from df_sample_aug$Treat   Format 1 2 1 1 1 1 2 2 2 2 2 2 1 1 2 2")
 print(Z_fsm_obs)


# Generate an assignment under CRD.
Z_crd = crd(df_sample, n_treat = 2, treat_sizes = c(n1, n2), control = FALSE)$Treat
print("Generate an assignment under CRD from df_sample - sway")
#print("printing df_sample ")
#print(df_sample)
print("Printing Z_crd from df_sample")
print(Z_crd)  #Z_crd printing in 1 2 1 2 2 etc format


# Generate an assignment under GSWD.
print("Generate a sample dataset from matrix(rnorm(840), nrow=120) - sway")
XGS_full <- matrix(rnorm(840), nrow=120) # Full sample with 120 rows and 7 columns 120 X 7
#print("Printing XGS_full - sway")
#print(XGS_full)
#print("---------")

print("Output from sample_gs_walk() using df_sample - Generates a single treatment assignment (TRUE or False) using the Gram–Schmidt Walk algorithm.")
#Z_gswd <- sample_gs_walk(XGS_full, phi = 0.3)  #initial
Z_gswd <- sample_gs_walk(as.matrix(df_sample[,-1]), phi = 0.3) #The function expected a covariate matrix. So dropped the index column from df_sample and converted it into matrix format
print("Z_gswd - the assignment - output from sample_gs_walk(as.matrix(df_sample[,-1]), phi = 0.3) in TRUE or FALSE Fformat")
print(Z_gswd) #Z_gswd printing in TRUE FALSE TRUE TRUE FALSE  etc format
#Convert output from TRUE/FALSE to 2/1  that is -  If TRUE make it 2, if FALSE make it 1
Z_gswd <- ifelse(Z_gswd, 2, 1)
print("Z_gswd - the assignment - output from sample_gs_walk(as.matrix(df_sample[,-1]), phi = 0.3) in 2 or 1  format")
print(Z_gswd)


 # Generate Love plot of the TASMDs in T_1 for FSM and CRD.
 print("Love plot (1) of the TASMDs in T_1 for FSM and CRD using df_sample")
 love_plot(data_frame = df_sample, index_col = TRUE, alloc1 = Z_fsm_obs, alloc2 = Z_crd, imbalance = 'TASMD', treat_lab = 1, legend_text = c('FSM','CRD'), xupper = 0.15)
 print("----------------------------------------------------------------")

 # Generate Love plot of the TASMDs in T_1 for FSM and GSWD. NEW
 print("Love plot (2) of the TASMDs in T_1 for FSM and GSWD using df_sample")
 love_plot(data_frame = df_sample, index_col = TRUE, alloc1 = Z_fsm_obs, alloc2 = Z_gswd, imbalance = 'TASMD', treat_lab = 1, legend_text = c('FSM','GSWD'), xupper = 0.15)
 print("----------------------------------------------------------------")


 # Check balance on squares and products of continuous covariates
 df_sample_sq_int = make_sq_inter(data_frame = df_sample[,c(2,3,6,7)], is_square = TRUE, is_inter = TRUE, keep_marginal = FALSE)  #Jiko Update

 print("Love plot (3) using TASMDs for df_sample_sq_int for FSM and CRD using df_sample_sq_int i.e., squares and products of continuous covariates")
 love_plot(data_frame = df_sample_sq_int, index_col = FALSE, alloc1 = Z_fsm_obs, alloc2 = Z_crd, imbalance = "TASMD", treat_lab = 1, legend_text = c('FSM','CRD'), xupper = 0.15)
 print("----------------------------------------------------------------")

 print("Love plot (4) using TASMDs for df_sample_sq_int for FSM and GSWD using df_sample_sq_int i.e., squares and products of continuous covariates")
 love_plot(data_frame = df_sample_sq_int, index_col = FALSE, alloc1 = Z_fsm_obs, alloc2 = Z_gswd, imbalance = "TASMD", treat_lab = 1, legend_text = c('FSM','GSWD'), xupper = 0.15)
 print("----------------------------------------------------------------")
```

```r
 # density function for 800 iteration
 # Generate 800 indep realizations of CRD & the FSM
 set.seed(9)

 # Set number of iterations
 n_iter = 800
 # Initialize assignment vectors for all the iterations
 Z_crd = matrix(rep(0,n_iter*N),nrow = n_iter)
 Z_re01 = matrix(rep(0,n_iter*N),nrow = n_iter)
 Z_re001 = matrix(rep(0,n_iter*N),nrow = n_iter)
 Z_fsm = matrix(rep(0,n_iter*N),nrow = n_iter)
 # gswd New
 Z_gswd = matrix(rep(0,n_iter*N),nrow = n_iter)

 print("Generate assignment under 5 Designs - CRD, rerandomization .01, rerandomization .001, FSM and GSWD")
 for (i in 1:n_iter) {

  ## Design-1: CRD




  fc = crd(df_sample, n_treat = 2, treat_sizes = c(n1,n2), control = FALSE)
  Z_crd[i,] = fc$Treat

    ## Design-2: Rerandomization tol 0.01
  re01 = rerand_emp(df_sample[,-1], tol = 0.01, is_prop = TRUE,
                    treat_sizes = c(n1,n2), control = FALSE, n_sim = 3000)
  Z_re01[i,] = re01$Treat

   ## Design-3: Rerandomization tol 0.001
  re001 = rerand_emp(df_sample[,-1], tol = 0.001, is_prop = TRUE,
                     treat_sizes = c(n1,n2), control = FALSE, n_sim = 3000)
  Z_re001[i,] = re001$Treat

  # --- Design 4: FSM (D-optimal) ---
  som_iter = som( data_frame = NULL, n_treat = 2, treat_sizes   = c(n1, n2),
        method = 'SCOMARS',  marginal_treat = rep((n2 / N), N), control = FALSE
         )
  f = fsm( data_frame  = df_sample, SOM = som_iter, s_function  = 'Dopt',
           eps = 0.0001,  ties = 'random', intercept = TRUE, standardize = TRUE,
           units_print = FALSE
         )
  Z_fsm[i, ] = f$data_frame_allocated$Treat

  # ---- Design-5 GSWD Gram Schmidt Walk Design ---------

  #new GSWD
  #fc_gswd = sample_gs_walk(df_sample, phi = 0.3)
  fc_gswd <- ifelse(sample_gs_walk(as.matrix(df_sample[,-1]), phi = 0.3), 2, 1)
  Z_gswd[i,] = fc_gswd
}
```

```r
# Diagnostics
print("Diagnostic begins")
ASMD_tab_original_1 = matrix(rep(0,k*4), nrow = k, ncol = 4)

ASMD_1 = asmd_rand(data_frame = df_sample, index_col = TRUE, alloc1 = Z_crd, alloc2 = Z_re01,
                     treat_lab = c(1,2), legend = c('CRD', 'RR 0.01'), roundoff = 3)

ASMD_2 = asmd_rand(data_frame = df_sample, index_col = TRUE, alloc1 = Z_re001, alloc2 = Z_fsm,
                     treat_lab = c(1,2), legend = c('RR 0.001', 'FSM'), roundoff = 3)

ASMD_tab_original_1[,1] = ASMD_1$asmd_mean[,1]
ASMD_tab_original_1[,2] = ASMD_1$asmd_mean[,2]
ASMD_tab_original_1[,3] = ASMD_2$asmd_mean[,1]
ASMD_tab_original_1[,4] = ASMD_2$asmd_mean[,2]

colnames(ASMD_tab_original_1) = c('CRD', 'RR 0.01', 'RR 0.001', 'FSM')
rownames(ASMD_tab_original_1) = colnames(X)

# New for GSWD added begin ----
ASMD_tab_original_3 = matrix(rep(0,k*4), nrow = k, ncol = 4)
ASMD_3 = asmd_rand(data_frame = df_sample, index_col = TRUE, alloc1 = Z_gswd, alloc2 = Z_fsm,
                     treat_lab = c(1,2), legend = c('GSWD', 'FSM'), roundoff = 3)
ASMD_tab_original_3[,1] = ASMD_3$asmd_mean[,1]
ASMD_tab_original_3[,2] = ASMD_3$asmd_mean[,2]
# New for GSWD added end ----


# Plots
# Concatenate the ASMDs across all the covariates
# Initialize vectors of ASMDs for both designs
ASMD_crd_iter = matrix(rep(0, n_iter*k), nrow = n_iter)
ASMD_re01_iter = matrix(rep(0, n_iter*k), nrow = n_iter)
ASMD_re001_iter = matrix(rep(0, n_iter*k), nrow = n_iter)
ASMD_fsm_iter = matrix(rep(0, n_iter*k), nrow = n_iter)

# New for GSWD added begin ----
ASMD_gswd_iter = matrix(rep(0, n_iter*k), nrow = n_iter)
# New for GSWD added end ----

for(r in 1:n_iter)
{
  ASMD_1 = asmd_rand(data_frame = df_sample, index_col = TRUE,
                       alloc1 = Z_crd[r,], alloc2 = Z_re01[r,],
                       treat_lab = c(1,2), legend = c('CRD','RR 0.01'), roundoff = 3)
  ASMD_crd_iter[r,] = ASMD_1[,1]
  ASMD_re01_iter[r,] = ASMD_1[,2]


  ASMD_2 = asmd_rand(data_frame = df_sample, index_col = TRUE,
                       alloc1 = Z_re001[r,], alloc2 = Z_fsm[r,],
                      treat_lab = c(1,2), legend = c('RR 0.001', 'FSM'), roundoff = 3)
  ASMD_re001_iter[r,] = ASMD_2[,1]
  ASMD_fsm_iter[r,] = ASMD_2[,2]


# New GSWD begin ---
  ASMD_3 = asmd_rand(data_frame = df_sample, index_col = TRUE,
                       alloc1 = Z_gswd[r,], alloc2 = Z_fsm[r,],
                      treat_lab = c(1,2), legend = c('GSWD', 'FSM'), roundoff = 3)
  ASMD_gswd_iter[r,] = ASMD_3[,1]
  ASMD_fsm_iter[r,] = ASMD_3[,2]
# New GSWD end ---

}

ASMD_crd_concat = as.vector(ASMD_crd_iter)
ASMD_re01_concat = as.vector(ASMD_re01_iter)
ASMD_re001_concat = as.vector(ASMD_re001_iter)
ASMD_fsm_concat = as.vector(ASMD_fsm_iter)
# New GSWD begin ---
ASMD_gswd_concat = as.vector(ASMD_gswd_iter)
# New GSWD end ---

t_crd_mean = round(mean(ASMD_crd_concat),3)
t_re01_mean = round(mean(ASMD_re01_concat),3)
t_re001_mean = round(mean(ASMD_re001_concat),3)
t_fsm_mean = round(mean(ASMD_fsm_concat),3)
# New GSWD begin ---
t_gswd_mean = round(mean(ASMD_gswd_concat),3)
# New GSWD end ---

# Plot estimated densities of the ASMDs under each design
print("Estimated densities of the ASMDs under each design, i.e. CRD, FSM, RR 0.01 and RR 0.001")
ASMD_df = data.frame(ASMD = c(ASMD_crd_concat,
                              ASMD_re01_concat,
                              ASMD_re001_concat,
                              ASMD_fsm_concat),
                      Design = rep(c(paste('CRD = ', t_crd_mean),
                                     paste('RR 0.01 = ', format(round(t_re01_mean, digits=3), nsmall = 3)),
                                     paste('RR 0.001 = ', t_re001_mean),
                                     paste('FSM = ', t_fsm_mean)),
                                   each = length(ASMD_crd_concat)))

ASMD_df$Design = factor(ASMD_df$Design, levels = unique(ASMD_df$Design))


# New GSWD begin ---
# Plot estimated densities of the ASMDs under FSM and GSWD only
print("Estimated densities of the ASMDs under two design i.e. FSM and GSWD")
ASMD_df2 = data.frame(ASMD = c(ASMD_gswd_concat,
                              ASMD_fsm_concat),
                      Design = rep(c(paste('GSWD = ', t_gswd_mean),
                                     paste('FSM = ', t_fsm_mean)),
                                   each = length(ASMD_gswd_concat)))

ASMD_df2$Design = factor(ASMD_df2$Design, levels = unique(ASMD_df2$Design))
# New GSWD end ---


# Plot
print("Density Plot - ASMD Main Covariates- mean ASMD")
p = ggplot(ASMD_df, aes(x = ASMD, colour = Design)) +
  geom_density(alpha = 0.5, adjust = 2.5, size = 1)
p + ylim(0,20) + labs(y = 'Density', x = 'ASMD Main Covariates') + labs(fill = "Mean ASMD") + theme_bw() +
  theme(panel.grid.major = element_blank(), panel.grid.minor = element_blank())+
  # scale_color_manual(values=rainbow(4)) +
  scale_fill_brewer(palette = "Set1") +
  scale_color_manual('Mean ASMD', values = c( "#FF7256", "#3CB371", "#40E0D0", "#EE82EE"))+
  theme(legend.position = c(0.73, 0.85), legend.text=element_text(size=15),
        text = element_text(size=15))

# New GSWD begin ---
# Plot for FS and GSWD only
print("Density Plot - ASMD Main Covariates- mean ASMD for FSM and GSWD")
p = ggplot(ASMD_df2, aes(x = ASMD, colour = Design)) +
  geom_density(alpha = 0.5, adjust = 2.5, size = 1)
p + ylim(0,20) + labs(y = 'Density', x = 'ASMD Main Covariates') + labs(fill = "Mean ASMD") + theme_bw() +
  theme(panel.grid.major = element_blank(), panel.grid.minor = element_blank())+
  # scale_color_manual(values=rainbow(4)) +
  scale_fill_brewer(palette = "Set1") +
  scale_color_manual('Mean ASMD', values = c( "#AA7256", "#EE82EE"))+
  theme(legend.position = c(0.73, 0.85), legend.text=element_text(size=15),
        text = element_text(size=15))
# New GSWD end ---
```

```r
# Computing similarity of covariance matrices
frob_crd = rep(0, n_iter)
frob_re01 = rep(0, n_iter)
frob_re001 = rep(0, n_iter)
frob_fsm = rep(0, n_iter)
# New GSWD begin -
frob_gswd = rep(0, n_iter)
# New GSWD end -

for(r in 1:n_iter)
{
  # CRD
  S1 = cor(X[Z_crd[r,] == 1,])
  S2 = cor(X[Z_crd[r,] == 2,])
  frob_crd[r] = norm((S1-S2), type = "F")
  #frob_crd[r] = norm((solve(S1)- solve(S2)), type = "F")

  # RR 0.01
  S1 = cor(X[Z_re01[r,] == 1,])
  S2 = cor(X[Z_re01[r,] == 2,])
  frob_re01[r] = norm((S1-S2), type = "F")
  #frob_re01[r] = norm((solve(S1)- solve(S2)), type = "F")

  # RR 0.001
  S1 = cor(X[Z_re001[r,] == 1,])
  S2 = cor(X[Z_re001[r,] == 2,])
  frob_re001[r] = norm((S1-S2), type = "F")
  #frob_re001[r] = norm((solve(S1)- solve(S2)), type = "F")


  # FSM Dopt
  S1 = cor(X[Z_fsm[r,] == 1,])
  S2 = cor(X[Z_fsm[r,] == 2,])
  frob_fsm[r] = norm((S1-S2), type = "F")
  #frob_fsm[r] = norm((solve(S1)- solve(S2)), type = "F")

# New GSWD begin -
  # GSWD
  S1 = cor(X[Z_gswd[r,] == 1,])
  S2 = cor(X[Z_gswd[r,] == 2,])
  frob_gswd[r] = norm((S1-S2), type = "F")
  #frob_fsm[r] = norm((solve(S1)- solve(S2)), type = "F")
# New GSWD end -

}
print("Frobenius Norm for CRD, RR 0.01, RR 0.001, FSM, GSWD")
par(mar = c(2, 4, 0.5, 1))
boxplot(frob_crd, frob_re01, frob_re001, frob_fsm, frob_gswd,
        ylim = c(0,2), cex.axis = 1.1, ylab = 'Frobenius Norm', cex.lab = 1.1,
        names = c('CRD', 'RR 0.01', 'RR 0.001', 'FSM', 'GSWD'))
abline(h=0, lty = "dashed")

imp = c((mean(frob_crd) - mean(frob_fsm))/mean(frob_crd),
        (mean(frob_re01) - mean(frob_fsm))/mean(frob_re01),
        (mean(frob_re001) - mean(frob_fsm))/mean(frob_re001),
        (mean(frob_gswd) - mean(frob_fsm))/mean(frob_gswd)
        )
```
